Pandas for ML

```python
# 1. Import the library (Pandas is usually called 'pd')
import pandas as pd

# 2. Create a simple dataset
data = {
    "Name": ["Amit", "Priya", "John", "Sara"],
    "Age": [22, 25, 23, 24],
    "City": ["Delhi", "Bangalore", "Mumbai", "Bangalore"],
    "Salary_LPA": [4.5, 6.0, 5.5, 7.0]
}

# 3. Convert it into a DataFrame (Excel-like table)
df = pd.DataFrame(data)

# 4. Print the table
print("--- The Table ---")
print(df)

# 5. Get basic info
print("\n--- Summary Statistics ---")
print(df.describe())

# 6. Filter data (Show only people from Bangalore)
print("\n--- People from Bangalore ---")
banglore_people = df[df['City'] == 'Bangalore']
print(banglore_people)
```
