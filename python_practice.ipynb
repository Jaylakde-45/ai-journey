```python
import pandas as pd
```

```python
df = pd.read_csv('electricity_data.csv')
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df = df.drop_duplicates()
print(df.shape)
```

```python
df['BillAmount'] = df['BillAmount'].fillna(df['BillAmount'].median())
```

```python
df.groupby('Area')['BillAmount'].agg('mean','sum','count')
```

```python
# Re-create df2 with a capital 'A'
df2 = pd.DataFrame({'Area': ['A', 'B', 'C'], 'region_head': ['North', 'West', 'South']})
```

```python
# Now run the merge again
merged = pd.merge(df, df2, on='Area', how='left')
merged.head()
```
