```python
import pandas as pd
import numpy as np

# 1. Create fake "dirty" data with missing values (NaN)
raw_data = {
    "Name": ["Amit", "Priya", "John", "Sara", "Rahul"],
    "Age": [22, np.nan, 23, 24, np.nan], # np.nan means missing number
    "Salary": [50000, 60000, np.nan, 70000, 45000],
    "City": ["Delhi", "Bangalore", "Mumbai", "Bangalore", "Delhi"]
}

df = pd.DataFrame(raw_data)

print("--- Original Data ---")
print(df)

# 2. Find the dirt (Missing values)
print("--- Missing Values per Column ---")
print(df.isnull().sum())

# 3. How to fix it? Fill missing Age with the average age!
average_age = df['Age'].mean()
print(f"\nAverage Age is: {average_age}")

# Fill the missing (NaN) values with the average
df['Age'] = df['Age'].fillna(average_age)

# Drop rows where Salary is missing (another strategy)
df = df.dropna(subset=['Salary'])

print("\n--- Cleaned Data ---")
print(df)
```

```python
import pandas as pd

# Load real data
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
titanic = pd.read_csv(url)

df = pd.DataFrame(titanic)

print(df)

# Deep Analysis: Who survived?
print("--- Survival Rate by Gender ---")
print(titanic.groupby("Sex")["Survived"].mean())

print("\n--- Survival Rate by Passenger Class ---")
print(titanic.groupby("Pclass")["Survived"].mean())
```
